## S7 · Verdict

The gates, the conclusion and the manifest are rendered below from the run's own objects.

```python
# ---- GATES: every gate as data; the table, S0 and the manifest render from it ----
from fundplat.quant.core import report as REPORT
GATES = []      # {"id": "G1", "section": "S3", "label": <24 chars, no number>, "name": ..., "reading": ..., "verdict": "PASS|FAIL|FLAG|INFO|N/A", "would_change": ...}
# THE FIXED CARD (the-study-manifest.md): fill PROFILE and the room shows this study in its fixed layout; with PROFILE set,
# every gate needs a label, the headline a label (<=16), every RULES value <=60 chars, verdicts from the closed set above
PROFILE = None  # {"markets": ["crypto|equities|rates|predmkts", ...] (the LABELS, main first; never the retired xasset or vol), "venues": [registry ids], "instruments": [...], "bar": "1h"}
BOOK = None     # {"at": "1x", "legs": [{"side": "LONG|SHORT|BUY|SELL", "instrument": ..., "account": ..., "usd": ...}]}; None for a study with no position
STYLE = None    # systematic (a fixed rule trades it) | discretionary (a held position the owner chose) | None for a pure evidence study; pace is READ from METRICS by build_manifest, never typed
FORMAT = None   # carry | directional | short_vol | rel_value | evidence: the ONE format label (the-study-manifest.md); needs METRICS
METRICS = None  # {key: number or None}, every key of FORMAT; a None names its reason in NA
NA = None       # {key: reason}
RULES = None    # {"entry", "exit", "sizing", "rebalance", "hedge", "hold", "stop", "cost"}: short strings, None where none
FIGURES = []    # EVERY cell that draws, by its fig:s<n>-<name> tag, in cell order: {"id", "section", "title", "kind"}, kind = the chart type's slug in app/library/research/charts.md; the checker compares
VERDICT = {"word": "DRAFT", "why": [], "next_bet": None}   # + "next": {"metric", "predict", "window"} when there is a next bet
CAVEATS = ["one window, one regime, one observation"]
# THE SUMMARY CARD (research.md#the-summary-card): the one fixed card the room and the last cell draw, built from the run's own objects
# BEFORE the manifest cell and carried in it. {"question", "answer", "implementation": [[name, fact] x 6], "big": [4 x {"scope", "metric",
# "value", "bar", "verdict"}], "blocks": [{"title", "rows": [{"metric", "value", "bar", "verdict"}]}], "sizes": {"cols", "rows", "chosen"},
# "footer": [1 to 4 lines]}. A value is a number (<= 26 chars), a verdict is PASS FAIL FLAG INFO or empty, and no code name shows. A row may add
# "key": a manifest metric its value must equal, or "gate": a gate id whose verdict it must carry (the validator then holds them). Counts of 6 and 4
# bind a carry card only. Draw it with fundplat.quant.core.summary_card.render(SUMMARY, carry=FORMAT == "carry").
SUMMARY = None
if GATES:
    display(Markdown(REPORT.gate_table_md(GATES)))
```

```python
# ---- the study manifest fund.study/1: built by fundplat's report helpers (validated on build),
# ---- emitted as ONE display output, and S0 filled from it ----
# the input pins: every series this kernel read through fundplat.data (the handle ledger), frozen as content-hashed pins; a read the
# store does not keep (a live book walk, a live quote) is named in `unpinned` with its reason, never dropped
from fundplat.data import handle as HANDLE
INPUTS = HANDLE.pin_reads(f"study input: {STUDY_ID}", f"study:{STUDY_ID}")
# INPUTS["unpinned"] += [{"what": "<feed>", "why": "<class none: stored nowhere>"}]     # one entry per live read
STUDY = REPORT.build_manifest(
    id=STUDY_ID, kind=KIND, desk=DESK, title=TITLE, question=QUESTION, status="draft",
    run={"executed": RUN_AT, "kernel": "ccvenv"}, inputs=INPUTS, window=WINDOW, verify=VERIFY, gates=GATES, figures=FIGURES,
    verdict=VERDICT, caveats=CAVEATS, rules=RULES, profile=PROFILE, book=BOOK,
    format=FORMAT, metrics=METRICS, na=NA, summary=SUMMARY, style=STYLE)
S0_HANDLE.update(Markdown(REPORT.summary_md(STUDY)))
display(REPORT.display_bundle(STUDY), raw=True)
```
