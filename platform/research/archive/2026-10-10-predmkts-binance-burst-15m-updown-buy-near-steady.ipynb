# Which variant of buying the near token a Binance burst left under fair value climbs steadily: the coin, the price band, one order a slot, the burst, the margin, the time left?

## S0 · Summary

**Research question.** Among the logical variants of the buy-near rule (all coins or Bitcoin; the near ask between 0.40 and 0.85, 0.60 and 0.85 or 0.70 and 0.90; one order a 15-minute slot across the coins or not; a 10 s burst of |z| at least 3 or 4; the ask 4 or 8 cents under the driftless fair value; at least 120 or 300 s left), held to expiry with one order per onset at a 250 ms delay, does the one with the steadiest climb on the validation days (the daily Sharpe, among cells above zero without their best 5% on both windows) earn a positive net PnL an entry on the test days with its 95% slot interval above zero, its net above zero without its best 5%, an equity line that fits a straight line, and a live replay that reproduces it?

**The idea.** Three grids on this rule (um-014, um-016, um-018) agree on three things: the rule is positive out of sample on 38 days (+1.25 USD an entry on 814, 30 of 38 days above zero, +0.33 without its best 5%), every exit is worth less than holding through the real ladder, and no layered signal survives a choice among hundreds of cells. What they also show is a lumpy climb: the money comes in days of many clustered bursts, the four coins burst together, and cheap entries pay large wins. The owner's criterion is a steady climb. The levers that change the shape without changing the idea are few: Bitcoin alone (the deepest book; +1.37 an entry out of sample against +0.14 for ETH and zero for SOL), a higher and narrower price band (smaller, more frequent wins), one order a slot across the coins (the four coins' bursts are one event), and the burst threshold, the margin and the time left that set how selective the rule is. 96 cells, chosen by the daily Sharpe on the validation days among cells above zero without their best 5% on both windows; a steadiness gate on the test days (the equity line's fit, the share of days above zero, the best 5%'s share of the net); the chosen cell replayed through its live form.
