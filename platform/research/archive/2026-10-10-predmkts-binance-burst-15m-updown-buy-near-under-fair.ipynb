# On every seated day, which setting of buying the near token a Binance burst left under-priced earns without a few wins carrying it?

## S0 · Summary

**Research question.** Does buying the near token of a Polymarket 15-minute Up/Down market after a Binance burst, only when its ask sits under the driftless fair value by a margin, with a price band on the ask, a minimum time left and an exit chosen on the train (09-02 to 09-22) and validation (09-23 to 10-02) days by the net an entry without the best 5%, earn a positive net PnL an entry on the test days 10-03 to 10-09 at a 250 ms delay, with its 95% slot interval above zero and its net above zero once its best 5% of entries are taken out?

**The idea.** um-014 (2026-10-10) searched these 810 settings on 22 days (train 09-16 to 09-28, validate 09-29 to 10-02, test 10-03 to 10-07) and chose L10|z4|th0.08|px40-85|T120|expiry: +1.49 USD an entry on 157 test entries, +0.54 without its best 5%, every robustness gate passed but the interval spanned zero and the deflated Sharpe was zero. Since then 16 more days seated (09-02 to 09-15, 10-08 and 10-09), and the owner's word (2026-10-11) is that every analysis runs on all 38. This run is the same grid on the 38-day split: 21 train days, 10 validation days, 7 test days, read at 250 ms (the venue's 150 ms hold plus 100 ms of our own, assumed), chosen by the net an entry WITHOUT the best 5% of entries so a cell that lives on tail wins cannot be chosen. um-016's filters and exits are not re-run here: um-018 read them on these 38 days (every exit worth less than holding, no signal family earning on both windows), and the seven-test notebook reads the non-momentum signals as fixed rules. The test days were inside um-011's whole-window read and um-021's seven fixed rules, so the cell choice, not the idea, is what is out of sample; the forward days from 10-10 are the clean read.
