# Which signals, added to the Binance burst, make buying the near token pay, and does any exit beat holding?

## S0 · Summary

**Research question.** Crossing a 10 s Binance burst (|z| 3 or 4) with one or two confirming signals of different families (a 30 s or 60 s trend, the window's level, Binance trade intensity, the book's imbalance, the spread, the Polymarket tape, the vol regime, RSI, the burst's first second, the time left, the UTC hour), with the fair-value gate off, at 4 cents or at 8 cents, on every coin or on Bitcoin alone, and with seven exits on the rules of at most one signal: does the cell chosen on 31 train and validation days by its net an entry without the best 5% earn on the 7 test days at a 250 ms delay, with its 95% slot interval above zero and its net above zero without its best 5%?

**The idea.** Four runs (um-011, um-014, um-016, um-017's design read) found the buy-near trade's point estimate above zero and its edge in a few wins on every coin pooled; um-016 tried filters one family at a time on um-014's cell and found none that held on both train and validation, with 4 validation days. Since then B-0684 seated 16 more days, so this grid has 19 train days and 12 validation days. The owner's ask (2026-10-10): explore many other signals, enter only when momentum and another signal agree, try logical combinations, and try exits, stop losses and take-profits. So the burst is the core of every cell, and each cell adds one or two confirming conditions from different families, the fair-value gate at three settings (off is the momentum-only form), two coin sets and, for the simpler rules, seven exits.

| | |
|---|---|
| Market | Polymarket 15-minute crypto Up/Down (BTC, ETH, SOL, XRP), driven by Binance spot trades |
| The trade | every second from the first decision to 5 s before the end, if the 10 s Binance z crosses the cell's threshold (3 or 4) in either direction, the near token's ask is between 0.40 and 0.85 (and under the driftless fair value by the cell's margin, when it has one), at least 120 s are left, and the cell's confirming signals hold, send one order for USD 20 of the near token, meeting the ladder 250 ms later at no worse than the ask seen at the decision; an order that does not fill is not sent again until the cell's condition has fallen and risen again (one order per signal onset, the owner's word 2026-10-10); hold to expiry, or leave on the cell's exit, sold into the bid ladder; one entry per window |
| Costs, always on | the market's own taker fee (rate x p x (1 - p), or the 20% maker rebate for a resting order), the exact ladder walked 150 ms after the decision, the venue's 5-share minimum |
| Money test | net USD per entry out of sample, with a 95% interval over 15-minute slots; win rate beside the share needed to break even; Sharpe; the deflated Sharpe for the number of cells tried |
| Days, in date order | train 09-02, 09-03, 09-04, 09-05, 09-06, 09-07, 09-08, 09-09, 09-10, 09-11, 09-12, 09-13, 09-14, 09-15, 09-16, 09-17, 09-18, 09-19, 09-20; validate 09-21, 09-22, 09-23, 09-24, 09-25, 09-26, 09-27, 09-28, 09-29, 09-30, 10-01, 10-02; test 10-03, 10-04, 10-05, 10-06, 10-07, 10-08, 10-09 (read once) |
